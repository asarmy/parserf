# FaultModelDataset

This notebook demonstrates `FaultModelDataset`, the data access layer for earthquake rupture
forecast datasets. Topics covered:

1. Loading a dataset
2. Exploring the data tables (`parent_ids`, `subsections`, `ruptures`, `rake_frequencies`)

```python
from parserf.models import FaultModel, FaultModelDataset
```

## 1. Loading a Dataset

Available fault models:
- `FaultModel.NSHMP_2023` — USGS NSHM CONUS v6.0.0
- `FaultModel.UCERF3_31` — UCERF3 fault model 3.1
- `FaultModel.UCERF3_32` — UCERF3 fault model 3.2

```python
ds = FaultModelDataset(FaultModel.NSHMP_2023)
print(ds)
```

## 2. Data Tables

### parent_ids

Maps parent fault names to integer IDs.

```python
print("Shape:", ds.parent_ids.shape)
ds.parent_ids.head()
```

### subsections

All fault subsections with geometry, computed dimensions, and parent fault info.

```python
print("Shape:", ds.subsections.shape)
print("Columns:", list(ds.subsections.columns))
ds.subsections.head()
```

### ruptures

Rupture scenarios with parsed subsection indices as sets of integers.

```python
print("Shape:", ds.ruptures.shape)
print("Columns:", list(ds.ruptures.columns))
ds.ruptures.head()
```

### rake_frequencies

Faulting style counts per parent fault, derived from rupture rake angles.

```python
print("Shape:", ds.rake_frequencies.shape)
print("Columns:", list(ds.rake_frequencies.columns))
ds.rake_frequencies.head(10)
```
